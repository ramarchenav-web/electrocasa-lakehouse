```python
from pyspark.sql import Row
tables_df = spark.sql("""
    SELECT table_schema, table_name
    FROM system.information_schema.tables
    WHERE table_catalog = 'electrocasa'
      AND table_schema <> 'information_schema'
""")

tables = tables_df.collect()

results = []
for row in tables:
    full_name = f"electrocasa.{row.table_schema}.{row.table_name}"
    try:
        cnt = spark.sql(f"SELECT COUNT(*) AS cnt FROM {full_name}").collect()[0][0]
    except Exception as e:
        cnt = -1
    results.append(Row(table_schema=row.table_schema, table_name=row.table_name, row_count=cnt))

display(spark.createDataFrame(results))
```

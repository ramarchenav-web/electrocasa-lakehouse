```sql
CREATE OR REPLACE TABLE electrocasa.gold.ventas_sucursal_mes_gold AS
WITH ventas_fechas AS (
    SELECT
        sucursal_id,
        monto_total,
        COALESCE(
            try_to_date(fecha_venta,'yyyy-MM-dd'),
            try_to_date(fecha_venta,'dd/MM/yyyy')
        ) AS fecha_venta_dt
    FROM electrocasa.silver.ventas_silver
)

SELECT
    sucursal_id,
    YEAR(fecha_venta_dt) AS anio,
    LPAD(
        CAST(MONTH(fecha_venta_dt) AS STRING),
        2,
        '0'
    ) AS mes,
    CONCAT(
        YEAR(fecha_venta_dt),
        '-',
        LPAD(CAST(MONTH(fecha_venta_dt) AS STRING),2,'0')
    ) AS periodo,
    SUM(monto_total) AS ventas_totales,
    ROUND(AVG(monto_total),2) AS ticket_promedio,
    COUNT(*) AS cantidad_ventas,
    CURRENT_TIMESTAMP() AS gold_generation_timestamp
FROM ventas_fechas
GROUP BY
    sucursal_id,
    YEAR(fecha_venta_dt),
    MONTH(fecha_venta_dt);
```

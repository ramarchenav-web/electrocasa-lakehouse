```sql
CREATE OR REPLACE TABLE electrocasa.gold.ranking_productos_gold AS

SELECT
    p.producto_id,
    p.nombre_producto,
    p.categoria,
    p.marca,
    SUM(v.cantidad) AS unidades_vendidas,
    ROUND(
        SUM(v.monto_total),
        2
    ) AS ventas_totales,
    ROUND(
        AVG(v.monto_total),
        2
    ) AS ticket_promedio,
    CURRENT_TIMESTAMP() AS gold_generation_timestamp
FROM electrocasa.silver.ventas_silver v
INNER JOIN electrocasa.silver.catalogo_productos_silver p
    ON v.producto_id = p.producto_id
GROUP BY
    p.producto_id,
    p.nombre_producto,
    p.categoria,
    p.marca;
```

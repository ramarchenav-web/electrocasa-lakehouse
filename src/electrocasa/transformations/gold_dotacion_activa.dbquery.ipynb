```sql
CREATE OR REPLACE TABLE electrocasa.gold.dotacion_activa_gold AS

WITH ultimo_evento AS (
    SELECT
        id_empleado,
        sucursal_id,
        tipo_evento,
        fecha_evento,
        ROW_NUMBER() OVER (
            PARTITION BY id_empleado
            ORDER BY fecha_evento DESC
        ) AS rn
    FROM electrocasa.silver.empleados_silver
)
SELECT
    sucursal_id,
    COUNT(*) AS cantidad_empleados_activos,
    CURRENT_TIMESTAMP() AS gold_generation_timestamp
FROM ultimo_evento
WHERE rn = 1
AND tipo_evento <> 'BAJA'
GROUP BY sucursal_id;
```

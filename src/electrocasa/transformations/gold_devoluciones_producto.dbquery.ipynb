```sql
CREATE OR REPLACE TABLE electrocasa.gold.devoluciones_producto_gold AS

SELECT
    p.producto_id,
    p.nombre_producto,
    p.categoria,
    p.marca,
    COUNT(*) AS cantidad_devoluciones,
    ROUND(
        SUM(d.monto_reembolso),
        2
    ) AS monto_total_reembolsado,
    ROUND(
        AVG(d.monto_reembolso),
        2
    ) AS promedio_reembolso,
    MAX(d.ingestion_timestamp) AS ultima_ingesta_fuente,
    CURRENT_TIMESTAMP() AS gold_generation_timestamp
FROM electrocasa.silver.devoluciones_silver d
INNER JOIN electrocasa.silver.catalogo_productos_silver p
    ON d.producto_id = p.producto_id
GROUP BY
    p.producto_id,
    p.nombre_producto,
    p.categoria,
    p.marca;
```

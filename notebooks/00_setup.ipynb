```python
%sql
CREATE CATALOG IF NOT EXISTS electrocasa;

CREATE SCHEMA IF NOT EXISTS electrocasa.bronze;
CREATE SCHEMA IF NOT EXISTS electrocasa.silver;
CREATE SCHEMA IF NOT EXISTS electrocasa.gold;

CREATE VOLUME IF NOT EXISTS electrocasa.bronze.vol_landing;
```

```python
%sql
-- Ingeniería
GRANT ALL PRIVILEGES
ON CATALOG electrocasa
TO `ingenieria`;

GRANT ALL PRIVILEGES
ON SCHEMA electrocasa.bronze
TO `ingenieria`;

GRANT ALL PRIVILEGES
ON SCHEMA electrocasa.silver
TO `ingenieria`;

GRANT ALL PRIVILEGES
ON SCHEMA electrocasa.gold
TO `ingenieria`;
```

```python
%sql
-- Acceso al catálogo
GRANT USE CATALOG
ON CATALOG electrocasa
TO `analistas`;

-- Acceso al schema GOLD
GRANT USE SCHEMA
ON SCHEMA electrocasa.gold
TO `analistas`;

-- Solo lectura sobre todas las tablas GOLD 
GRANT SELECT
ON SCHEMA electrocasa.gold
TO `analistas`;
```

```python
%sql
GRANT USE CATALOG
ON CATALOG electrocasa
TO `auditoria`;

GRANT USE SCHEMA
ON SCHEMA electrocasa.gold
TO `auditoria`;

GRANT SELECT
ON SCHEMA electrocasa.gold
TO `auditoria`;
```

```python
# Catálogo
spark.sql("CREATE CATALOG IF NOT EXISTS electrocasa")

# Schemas
spark.sql("CREATE SCHEMA IF NOT EXISTS electrocasa.bronze")
spark.sql("CREATE SCHEMA IF NOT EXISTS electrocasa.silver")
spark.sql("CREATE SCHEMA IF NOT EXISTS electrocasa.gold")
spark.sql("CREATE SCHEMA IF NOT EXISTS electrocasa.auditoria")

# Volume principal
spark.sql("""
CREATE VOLUME IF NOT EXISTS electrocasa.bronze.vol_landing
""")

# Ventas
dbutils.fs.mkdirs("/Volumes/electrocasa/bronze/vol_landing/ventas_sucursales")

# Catálogo
dbutils.fs.mkdirs("/Volumes/electrocasa/bronze/vol_landing/catalogo_productos")

# Empleados (Auto Loader)
dbutils.fs.mkdirs("/Volumes/electrocasa/bronze/vol_landing/empleados_rrhh/data")
dbutils.fs.mkdirs("/Volumes/electrocasa/bronze/vol_landing/empleados_rrhh/schema")
dbutils.fs.mkdirs("/Volumes/electrocasa/bronze/vol_landing/empleados_rrhh/checkpoint")

# Reseñas
dbutils.fs.mkdirs("/Volumes/electrocasa/bronze/vol_landing/resenas_clientes")

# Devoluciones (Auto Loader)
dbutils.fs.mkdirs("/Volumes/electrocasa/bronze/vol_landing/devoluciones/data")
dbutils.fs.mkdirs("/Volumes/electrocasa/bronze/vol_landing/devoluciones/schema")
dbutils.fs.mkdirs("/Volumes/electrocasa/bronze/vol_landing/devoluciones/checkpoint")
```

```python
%sql
-- Ingeniería
GRANT ALL PRIVILEGES
ON CATALOG electrocasa
TO `ingenieria`;

GRANT ALL PRIVILEGES
ON SCHEMA electrocasa.bronze
TO `ingenieria`;

GRANT ALL PRIVILEGES
ON SCHEMA electrocasa.silver
TO `ingenieria`;

GRANT ALL PRIVILEGES
ON SCHEMA electrocasa.gold
TO `ingenieria`;

GRANT READ VOLUME
ON VOLUME electrocasa.bronze.vol_landing
TO `ingenieria`;

GRANT WRITE VOLUME
ON VOLUME electrocasa.bronze.vol_landing
TO `ingenieria`;
```

```python
%sql
GRANT USE CATALOG
ON CATALOG electrocasa
TO `analistas`;

GRANT USE SCHEMA
ON SCHEMA electrocasa.gold
TO `analistas`;

GRANT SELECT
ON SCHEMA electrocasa.gold
TO `analistas`;
```
